```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
```

```python
filename = "G:/neural_network_from_scratch/heart diseases/heart.csv"
df = pd.read_csv(filename)
df.head(1)
```

```python
sorted_df = df[['Age','Sex','ChestPainType','RestingBP','ExerciseAngina','Oldpeak','ST_Slope','HeartDisease']]
sorted_df.head()
```

```python
categorical_cols = [
    'Sex',
    'ChestPainType',
    'ExerciseAngina',
    'ST_Slope'
]

df_enc = pd.get_dummies(
    sorted_df,
    columns=categorical_cols,
    dtype=int
)

df_enc.head()
```

```python
X = df_enc.drop(['HeartDisease'],axis=1)
Y = df_enc['HeartDisease']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y,
    test_size=0.2,
    random_state=42,
    stratify=Y
)
```

```python
print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
```

```python
X_train
```

```python
# standardizing 
mean = np.mean(X_train, axis=0)
std = np.std(X_train, axis=0)

X_train = (X_train - mean)/std
X_test = (X_test - mean)/std
```

```python
X_train.describe()
```

```python
print("X_train mean:", np.mean(X_train, axis=0))
print("X_train std:", np.std(X_train, axis=0))

print("X_train max:", np.max(np.abs(X_train)))
```

```python
Y_train = Y_train.to_numpy().reshape(-1, 1)
Y_test = Y_test.to_numpy().reshape(-1, 1)
```

```python
# weights and bias
weights = np.zeros((X_train.shape[1],1))
bias = 0.0
```

```python
def sigmoid(z):
    return 1/(1+np.exp(-z))

print(sigmoid(2))
```

```python
def forward(X,weights,bias):
    z = np.dot(X,weights) + bias
    y_pred = sigmoid(z)
    return y_pred
```

```python
def loss_function(y,y_pred):
    epsilon = 1e-15
    y_pred = np.clip(y_pred,epsilon,1-epsilon)
    loss = -np.mean(y*np.log(y_pred)+(1-y)*np.log(1-y_pred))
    return loss
```

```python
y_pred = forward(X_train, weights, bias)
loss = loss_function(Y_train, y_pred)
print("Initial Loss:", loss)
```

```python
def gradient(X, y, y_pred, weights, lambda_reg):
    m = X.shape[0]
    error = y_pred - y
    dw = np.dot(X.T, error) / m
    db = np.sum(error) / m
    # L2 regularization
    dw += (lambda_reg / m) * weights
    return dw, db
```

```python
def gradient_descent(weights,bias,dw,db,learning_rate):
    weights = weights - learning_rate * dw
    bias = bias - learning_rate * db
    return weights, bias
```

```python
learning_rate = 0.1
epochs = 5000
epoch_list = [n for n in range(1,epochs+1)]
loss_history = []

for epoch in range(epochs):
    y_pred = forward(X_train,weights,bias)
    loss = loss_function(Y_train,y_pred)
    loss_history.append(loss)
    dw, db = gradient(X_train,Y_train,y_pred,weights,lambda_reg=0.1)
    weights,bias = gradient_descent(weights,bias,dw,db,learning_rate)
    if epoch%500 == 0 :
        print(f"epoch: {epoch}, loss: {loss}")
```
